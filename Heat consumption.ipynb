```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
year = np.array([2013,2014,2015,2016,2017,2018,2019,2020,2021,2022])
hc_glasgow = np.array([5547,5598,5433,5251,5403,5453,5457,5556,5445,5013])
hc_edinburgh = np.array([4916,4826,4759,4751,4924,4916,4900,4875,4808,4379])
hc_highlands = np.array([2652,2691,2711,2675,2743,2610,2636,2735,2967,3366])

norm_glasgow = hc_glasgow / np.mean(hc_glasgow)
norm_edinburgh = hc_edinburgh / np.mean(hc_edinburgh)
norm_highlands = hc_highlands / np.mean(hc_highlands)

pop_glasgow = 622050
pop_edinburgh = 514990
pop_highlands = 235710

fig, (ax) = plt.subplots(1, 1, figsize=(10, 6))
plt.rc('font',family='serif')
plt.rcParams.update({'font.size': 14})

ax.plot(year, norm_glasgow, label="Glasgow", color="hotpink")
ax.plot(year, norm_edinburgh, label="Edinburgh", color="purple")
ax.plot(year, norm_highlands, label="Highlands", color="darkblue")
ax.set_xticks(year)
ax.set_xlabel("Year")
ax.set_ylabel("Mean Normalised Fuel Consumption (GWh)")
ax.legend()

plt.tight_layout()
plt.savefig("fuel_consumption.pdf")
plt.show()
```
