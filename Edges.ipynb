```python
from PIL import Image

red = (237, 28, 36)       # Red
orange = (255, 127, 39)   # Orange

# load the image
img = Image.open("cleaned_image_faults.tiff").convert("RGB")
pixels = img.load()

w, h = img.size

# first, collect all red pixel positions
red_pixels = []
for y in range(h):
    for x in range(w):
        if pixels[x, y] == red:
            red_pixels.append((x, y))

# process neighbors
for x, y in red_pixels:
    neighbors = [
        (x, y - 1),  # above
        (x, y + 1),  # below
        (x - 1, y),  # left
        (x + 1, y)   # right
    ]
    for nx, ny in neighbors:
        if 0 <= nx < w and 0 <= ny < h:
            if pixels[nx, ny] != red:
                pixels[nx, ny] = orange

# save the updated image
img.save("image_faults.tiff")
```
