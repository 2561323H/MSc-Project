Calculate heat rate values

```python
# Baseline values
rho = 2800
Ck2o_wt = 2.1
Cu_ppm = 2
Cth_ppm = 8.4

Cu_wt = Cu_ppm/10000
```

```python
A_base = 10**(-5)*rho*(3.5*Ck2o_wt + 9.67*Cu_wt + 2.63*Cth_ppm)
print('Baseline heat rate of the gneiss is {0} W/m3'.format(A_base*10**(-6)))
```

```python
#Minimum values
rho = 2800
Ck2o_wt = 0.5
Cu_ppm = 1.7
Cth_ppm = 3

Cu_wt = Cu_ppm/10000

A_min = 10**(-5)*rho*(3.5*Ck2o_wt + 9.67*Cu_wt + 2.63*Cth_ppm)
print('Minimum heat rate of the gneiss is {0} W/m3'.format(A_min*10**(-6)))
```

```python
#Maximum values
rho = 2800
Ck2o_wt = 4
Cu_ppm = 3
Cth_ppm = 12

Cu_wt = Cu_ppm/10000

A_max = 10**(-5)*rho*(3.5*Ck2o_wt + 9.67*Cu_wt + 2.63*Cth_ppm)
print('Maximum heat rate of the gneiss is {0} W/m3'.format(A_max*10**(-6)))
```

Calculate rayleigh number

```python
g = 9.81
aT = 2.5e-5
kp = 1e-13
D = 5000
mu = 1e-3
dT = 47/1000*5000-8.2
rho = 1000
Cp = 4000
sigma = 2.5

RA = (g*aT*dT*kp*D*rho*Cp)/(sigma*mu)
print(RA)
```
