```python
from PIL import Image
import math

# define target colors
allowed_colors = {
    (255, 230, 128),  # #FFE680
    (255, 255, 255)   # #FFFFFF
}

# load the image
img = Image.open("image.tiff").convert("RGB")
pixels = img.load()

# function to calculate Euclidean distance between two RGB colors
def color_distance(c1, c2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(c1, c2)))

# find the nearest allowed color
def get_nearest_color(color, allowed_colors):
    return min(allowed_colors, key=lambda c: color_distance(color, c))

# process pixels
for y in range(img.height):
    for x in range(img.width):
        current_color = pixels[x, y]
        if current_color not in allowed_colors:
            pixels[x, y] = get_nearest_color(current_color, allowed_colors)

# save result
img.save("cleaned_image.tiff")
```
